```python
import requests

def route_afstand_osrm(punt1, punt2, profiel="driving"):
    """
    punt1, punt2: tuples van (lat, lon)
    profiel: "driving", "walking" of "cycling"
    Retourneert afstand in km en reistijd in minuten.
    """
    lon1, lat1 = punt1[1], punt1[0]
    lon2, lat2 = punt2[1], punt2[0]

    url = f"http://router.project-osrm.org/route/v1/{profiel}/{lon1},{lat1};{lon2},{lat2}"
    params = {"overview": "false"}

    response = requests.get(url, params=params)
    data = response.json()

    if data["code"] != "Ok":
        raise Exception(f"Route niet gevonden: {data}")

    afstand_km = data["routes"][0]["distance"] / 1000
    tijd_min = data["routes"][0]["duration"] / 60

    return afstand_km, tijd_min
```

```python
import pandas as pd

resultaten = []

for latj, lonj, naamj in totale_groep:
    if naamj in ("Utrecht WH", "Den Hoorn WH", "Haarlem WH", "Amsterdam WH") or naamj in winkels:
        for lati, loni, naami in totale_groep:
            if naami in ("Utrecht WH", "Den Hoorn WH", "Haarlem WH", "Amsterdam WH") or naami in winkels:
                afstand, tijd = route_afstand_osrm((lati, loni), (latj, lonj))
                if afstand < 75:
                    resultaten.append({
                        "Locatie1": naamj,
                        "Locatie2": naami,
                        "afstand": round(afstand,1),
                        "tijd": tijd,
                    })

df = pd.DataFrame(resultaten)

# Tabel: winkels (naami) als rijen, distributiecentra (naamj) als kolommen
tabel_afstand = df.pivot_table(index="Locatie1", columns="Locatie2", values="afstand")
tabel_tijd = df.pivot_table(index="Locatie1", columns="Locatie2", values="tijd")

print(tabel_afstand)
                
```
